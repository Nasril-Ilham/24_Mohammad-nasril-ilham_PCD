````python
# Tugas 3 Operasi Matematika

# Minta pengguna memasukkan dua angka.
# Konversikan input menjadi `int` atau `float`.
# Hitung penjumlahan, pengurangan, perkalian, pembagian, dan pangkat.
# Tampilkan hasil setiap operasi.
# Tambahkan perhitungan luas persegi panjang berdasarkan input panjang dan lebar.
````

````python
# variabel untuk menerima input 1
input1 = input("Masukkan angka pertama: ")

#variabel untuk menerima input 2
input2 = input("Masukkan angka kedua: ")

# karna input string dan kita akan melakukan aritmatika maka kita perlu menguabh tipe datana ke float / int
# aritmatika penjumlahan
hasil_penjumlahan = int(input1) + int(input2)
# artiamtika pengurangan
hasil_pengurangan = int(input1) - int(input2)
# aritmatika perkalian
hasil_perkalian = int(input1) * int(input2)
# aritamtika pembagian
hasil_pembagian = int(input1) / int(input2)
# aritmatika pangkat
hasil_pangkat = float(input1) ** float(input2)


# output dari aritamtika di atas
print()
print("Hasil penjumlahan:", hasil_penjumlahan)
print("Hasil pengurangan:", hasil_pengurangan)
print("Hasil perkalian:", hasil_perkalian)
print("Hasil pembagian:", hasil_pembagian)
print("Hasil pangkat:", hasil_pangkat)
print()
````

````python
# Tugas 4


# Buat program untuk menampilkan data mahasiswa dengan ketentuan:

# Buat daftar minimal lima nama mahasiswa.
# Tampilkan semua nama menggunakan for.
# Tampilkan angka 1 sampai 10 menggunakan range().
# Tampilkan kembali daftar nama menggunakan while dan len().
# Buat perulangan while untuk menampilkan bilangan genap dari 2 sampai 20.
````

````python
# variabel nama mahasiswa
daftar_mahasiswa = ['nasril', 'fikri', 'firman', 'saif', 'rijal']

# menampilkan daftar mahasiswa menggunakan for
print("--- Daftar Mahasiswa (for loop) ---")
for mahasiswa in daftar_mahasiswa:
    print(mahasiswa)

print()
print('------------')
print()

# menampilkan angka dari 1 sampai 10
print("--- Angka 1 sampai 10 ---")
for angka in range(1, 11):
    print(angka)

print()
print('------------')
print()

# menampilkan daftar mahasiswa menggunakan while dan len()
print("--- Daftar Mahasiswa (while loop & len) ---")
index = 0
while index < len(daftar_mahasiswa):
    print(daftar_mahasiswa[index])
    index += 1

print()
print('------------')
print()

# perulangan while untuk menampilkan bilangan genap dari 2 sampai 20
print("--- Bilangan Genap 2 sampai 20 ---")
bilangan = 2
while bilangan <= 20:
    print(bilangan)
    bilangan += 2
````

````python
# Tugas 5 — Fungsi dan Waktu

# Buat program pencatatan aktivitas belajar dengan ketentuan:

# - Impor modul `datetime`.
# - Buat fungsi bernama `catat_waktu()`.
# - Tampilkan pesan `Aktivitas selesai`.
# - Tampilkan waktu saat fungsi dijalankan menggunakan `datetime.datetime.now()`.
# - Panggil fungsi sebelum dan sesudah perulangan.
# - Gunakan perulangan `for` untuk menampilkan proses pengerjaan lima soal.
````

````python
import datetime

# Membuat fungsi catat_waktu()
def catat_waktu():
    print("Aktivitas selesai pada:", datetime.datetime.now())

# Memanggil fungsi sebelum perulangan
print("--- Memulai Aktivitas Belajar ---")
catat_waktu()
print()

# Menggunakan perulangan for untuk menampilkan proses pengerjaan lima soal
for i in range(1, 6):
    print(f"Sedang mengerjakan soal nomor {i}...")

print()
# Memanggil fungsi sesudah perulangan
print("--- Mengakhiri Aktivitas Belajar ---")
catat_waktu()
````

````python
# # Tugas proyek akhir — Sistem Rekap Nilai Mahasiswa
#
# Gabungkan seluruh materi dalam satu program sederhana dengan ketentuan berikut.
#
# Studi kasus
#
# Sebuah program digunakan untuk mencatat data mahasiswa, menerima nilai, menghitung rata-rata, dan menampilkan waktu penyelesaian proses.
#
# Spesifikasi program:
#
# 1. Tampilkan judul program menggunakan `print()` dan garis pemisah.
# 2. Minta input nama dan NIM mahasiswa.
# 3. Minta pengguna memasukkan nilai tiga mata kuliah.
# 4. Hitung total dan rata-rata nilai menggunakan operasi matematika.
# 5. Tampilkan nama mahasiswa dalam format huruf kapital menggunakan `upper()`.
# 6. Gunakan `for` untuk menampilkan seluruh nilai.
# 7. Gunakan `while` untuk menampilkan nomor urut mata kuliah.
# 8. Buat fungsi `tampilkan_hasil()` untuk menampilkan total dan rata-rata.
# 9. Buat fungsi `catat_waktu()` untuk menampilkan waktu saat program menyelesaikan proses.
# 10. Tampilkan hasil akhir secara rapi dan mudah dibaca.
#
# ### Contoh gambaran output proyek akhir
#
# ```
# ================================
#        REKAP NILAI MAHASISWA
# ================================
# Nama       : ANDI PRATAMA
# NIM        : 123456
# --------------------------------
# Nilai Algoritma   : 80
# Nilai Basis Data  : 85
# Nilai Struktur Data: 90
# --------------------------------
# Total Nilai : 255
# Rata-rata   : 85.0
# --------------------------------
# Status      : Rekap selesai
# Waktu       : [waktu program dijalankan]
# ================================
# ```
````

````python
import datetime

# 9. Buat fungsi catat_waktu() untuk menampilkan waktu saat program menyelesaikan proses
def catat_waktu():
    return datetime.datetime.now()

# 8. Buat fungsi tampilkan_hasil() untuk menampilkan total dan rata-rata
def tampilkan_hasil(total, rata_rata):
    print(f"Total Nilai : {total}")
    print(f"Rata-rata   : {rata_rata:.1f}")

# 1. Tampilkan judul program menggunakan print() dan garis pemisah
print("====================================")
print("       REKAP NILAI MAHASISWA        ")
print("====================================")

# 2. Minta input nama dan NIM mahasiswa
nama_mhs = input("Nama       : ")
nim_mhs = input("NIM        : ")

# Daftar mata kuliah yang akan dinilai
mata_kuliah = ["Algoritma", "Basis Data", "Struktur Data"]
nilai_mhs = []

# 7. Gunakan while untuk menampilkan nomor urut mata kuliah
# Minta pengguna memasukkan nilai tiga mata kuliah
index = 0
while index < len(mata_kuliah):
    nilai = float(input(f"Masukkan Nilai {index + 1} ({mata_kuliah[index]}): "))
    nilai_mhs.append(nilai)
    index += 1

# 4. Hitung total dan rata-rata nilai menggunakan operasi matematika
total_nilai = sum(nilai_mhs)
rata_rata = total_nilai / len(nilai_mhs)

# 10. Tampilkan hasil akhir secara rapi dan mudah dibaca
print("\n====================================")
print("       REKAP NILAI MAHASISWA        ")
print("====================================")
# 5. Tampilkan nama mahasiswa dalam format huruf kapital menggunakan upper()
print(f"Nama       : {nama_mhs.upper()}")
print(f"NIM        : {nim_mhs}")
print("------------------------------------")

# 6. Gunakan for untuk menampilkan seluruh nilai
for i in range(len(mata_kuliah)):
    print(f"Nilai {mata_kuliah[i]:<13} : {nilai_mhs[i]}")

print("------------------------------------")
# Memanggil fungsi tampilkan_hasil()
tampilkan_hasil(total_nilai, rata_rata)
print("------------------------------------")
print("Status      : Rekap selesai")
# Memanggil fungsi catat_waktu()
print(f"Waktu       : {catat_waktu()}")
print("====================================")
````
