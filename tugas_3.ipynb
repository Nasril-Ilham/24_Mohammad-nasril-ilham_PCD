```python
# import cv2
import cv2

# memanggil fungsi google colab untuk perbaikan cv2.imshow
from google.colab.patches import cv2_imshow

# menyimpan gambar dengan fungsi imread dari OpenCV
img = cv2.imread("orange.png")

# menampilkan gambar dengan fungsi cv2_imshow
cv2_imshow(img)

print()

# lihat tipe data img. disimpan sebagai apa?
print(type(img))
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

try:
    # Membuka gambar orange.png dan mengubahnya menjadi array NumPy
    im = np.array(Image.open("orange.png"))

    # Memisahkan kanal Merah, Hijau, dan Biru

    # Menyalin gambar asli untuk pengolahan kanal merah
    im_R = im.copy()
    im_R[:, :, (1, 2)] = 0  # Menghilangkan kanal Hijau dan Biru

    # Menyalin gambar asli untuk pengolahan kanal hijau
    im_G = im.copy()
    im_G[:, :, (0, 2)] = 0  # Menghilangkan kanal Merah dan Biru

    # Menyalin gambar asli untuk pengolahan kanal biru
    im_B = im.copy()
    im_B[:, :, (0, 1)] = 0  # Menghilangkan kanal Merah dan Hijau

    # Menggabungkan ketiga gambar secara horizontal (axis=1)
    im_RGB = np.concatenate((im_R, im_G, im_B), axis=1)

    # Mengubah array NumPy menjadi objek gambar Pillow
    pil_img = Image.fromarray(im_RGB)

    # Menyimpan hasil pemisahan kanal ke file PNG
    pil_img.save("orange_channels_result.png")

    # Membuat area gambar berukuran 6 x 6 inci
    # karna kita di minta untuk ukutan 6x6 maka figsize nya jadi kecil
    plt.figure(figsize=(6, 6))

    # Menampilkan gambar hasil penggabungan kanal
    plt.imshow(pil_img)

    # Memberikan judul pada visualisasi
    plt.title("Pemisahan Kanal RGB (Red | Green | Blue)")

    # Menyembunyikan sumbu koordinat agar gambar lebih jelas
    plt.axis("off")

    # Menampilkan hasil visualisasi
    plt.show()
except Exception as e:
    print("Terjadi kesalahan:", e)
```

Soal 2 (NumPy - Cropping):

Lakukan pemotongan (cropping) gambar pada koordinat baris/y  200  hingga  600  dan kolom/x  400  hingga  800  menggunakan slicing array NumPy, lalu tampilkan hasilnya

```python
# Melakukan cropping menggunakan slicing NumPy
# Format slicing pada array 2D/3D: [y_start:y_end, x_start:x_end]
# im disini mengambil foto dari di atas yang sudah di ubah ke array jadi gambar di ambil dari array atas lalu di cropping
crop_img = im[200:600, 400:800]

# Menampilkan hasil crop dengan Matplotlib
plt.figure(figsize=(6, 6))
plt.imshow(crop_img)
plt.title("Hasil Cropping Gambar")
plt.axis("off")
plt.show()
```

**Soal 3 (Scikit-Image - Filtering):**
*   Konversi hasil potongan (crop) dari Soal 2 menjadi grayscale, lalu terapkan filter deteksi tepi Sobel (`filters.sobel`) dari Scikit-Image dan tampilkan hasilnya dengan peta warna (cmap) abu-abu (`gray`).

```python
## Jawaban Soal No 3
```

```python
from skimage import color
import matplotlib.pyplot as plt

# Mengonversi hasil crop (crop_img) menjadi grayscale
# karna di atas sudah di crop dengan nama variabel (crop_img) dan kita mau melakukan perubahan pada potongn maka kita panggil yang sudah di potong lalu di ubah ke gray scale
gray_crop = color.rgb2gray(crop_img)

# Menampilkan hasil grayscale dengan Matplotlib menggunakan cmap='gray'
plt.figure(figsize=(6, 6))
plt.imshow(gray_crop, cmap='gray')
plt.title("Hasil Konversi Grayscale")
plt.axis("off")
plt.show()
```

```python
from skimage import filters
import matplotlib.pyplot as plt

# Menerapkan filter deteksi tepi Sobel
# konsepnya sama seperti di atas untuk gambarnya ambil dari mana
edge_sobel = filters.sobel(gray_crop)

# Menampilkan hasil deteksi tepi Sobel
plt.figure(figsize=(6, 6))
plt.imshow(edge_sobel, cmap='gray')
plt.title("Hasil Deteksi Tepi Sobel")
plt.axis("off")
plt.show()
```
